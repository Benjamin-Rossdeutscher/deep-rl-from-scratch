### Introduction to PyTorch Tensors: from the PyTorch documentation/tutorials

```python
x = torch.empty(3,4)
print(type(x))
print(x)
```

```python
#factory methods to create tensors
zeros = torch.zeros(3,4)        #empty tensor of size 3x4 filled with zeros
print(zeros)
ones = torch.ones(3,4)          #empty tensor of size 3x4 filled with ones
print(ones)

torch.manual_seed(123)  #set the random seed for reproducibility
random = torch.rand(3,4)            #empty tensor of size 3x4 filled with random numbers
print(random)
```

### Random Seeding

```python
torch.manual_seed(1234)             #set the random seed for reproducibility
random1 = torch.rand(3,4)           #demonstrate that the random numbers generated are the same when the seed is set to the same value
print(random1)

random2 = torch.rand(3,4)           #demonstrate that the following rng numbers hold after the reset
print(random2)

torch.manual_seed(1234)
random3 = torch.rand(3,4)           #reproduction of random1
print(random3)

random4 = torch.rand(3,4)           #reproduction of random2
print(random4)
```
