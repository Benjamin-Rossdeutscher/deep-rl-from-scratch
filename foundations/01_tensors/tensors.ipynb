### Random Seeding

```python
torch.manual_seed(1234)             #set the random seed for reproducibility
random1 = torch.rand(3,4)           #demonstrate that the random numbers generated are the same when the seed is set to the same value
print(random1)

random2 = torch.rand(3,4)           #demonstrate that the following rng numbers hold after the reset
print(random2)

torch.manual_seed(1234)
random3 = torch.rand(3,4)           #reproduction of random1
print(random3)

random4 = torch.rand(3,4)           #reproduction of random2
print(random4)
```

### Tensor Shapes

```python
x = torch.empty(1,2,3)
print(x.shape)
print(x)


empty_like_x = torch.empty_like(x)  #create an empty tensor with the same shape as x
print(empty_like_x.shape)
print(empty_like_x)

zeros_like_x = torch.zeros_like(x)
print(zeros_like_x.shape)
print(zeros_like_x)

ones_like_x = torch.ones_like(x)
print(ones_like_x.shape)
print(ones_like_x)

rand_like_x = torch.rand_like(x)
print(rand_like_x.shape)
print(rand_like_x)
```

### Math and Logic with PyTorch Tensors

Scalar operations and exponentiation act on each element individually.

```python
ones = torch.zeros(2, 2) + 1
twos = torch.ones(2, 2) * 2
threes = (torch.ones(2, 2) * 7 - 1) / 2
fours = twos**2
sqrt2s = twos**0.5

print(ones)
print(twos)
print(threes)
print(fours)
print(sqrt2s)
```

When applying BIMDAS operations between two tensors of equal size, each corresponding element interacts with its counterpart and maps into a new tensor of equal dimension.

```python
powers2 = twos ** torch.tensor([[1, 2], [3, 4]])
print(powers2)

fives = ones + fours
print(fives)

dozens = threes * fours
print(dozens)
```

### Tensor Broadcasting

The rules for broadcasting are:

1. Each tensor must have at least one dimension - no empty tensors.

2. Comparing the dimension sizes of the two tensors, going from last to first:

3. Each dimension must be equal, or

4. One of the dimensions must be of size 1, or

5. The dimension does not exist in one of the tensors

The smaller tensor is then 

```python
broadtest = torch.ones(1,2,2)
print(broadtest)
```

Ok so essentially a dimension with extent of 0 must be empty, as actually any non empty tensor has extent of 1 in every dimension, think of it as a 1D vector is a row/column in a matrix, ie. the plane is one point/element in a cube, it possesses existence that can be broadcasted or scaled into any higher dimension, which is what is happening here and why extent of 1 or not named are treated equivalently. Since it isn't defined which row/column the vector is in the dimensional extent, and broadcasting is done equally to each point in the new dim, each 1 is scaled to the larger value, if the no. dimensions don't match, PyTorch adds ones to the left till they do, so you match from the right to left.

```python
twodof = torch.ones(2,3,4)
print(twodof*torch.ones(2,1,1)*2)
```

### Manipulating Tensor Shapes

```python
a = torch.ones(1,2,1,2)
print(a.shape)
b = a.squeeze(-4).shape
print(b)
```

```python
a = torch.ones(4, 3, 2)
print(a.shape)
b = torch.rand(3)  # trying to multiply a * b will give a runtime error
c = b.unsqueeze(1)  # change to a 2-dimensional tensor, adding new dim at the end
print(c.shape)
print(a * c)  # broadcasting works again!
```

```python
output3d = torch.rand(6, 20, 20)
print(output3d.shape)

input1d = output3d.reshape(6 * 20 * 20)
print(input1d.shape)

# can also call it as a method on the torch module:
print(torch.reshape(output3d, (6 * 20 * 20,)).shape)
```
